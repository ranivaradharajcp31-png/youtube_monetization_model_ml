## Initial observations

- 122,400 rows, 12 columns. Target: `ad_revenue_usd`.
- Three columns have missing values: `likes`, `comments`, `watch_time_minutes` (~5% each).
- No negative or zero values anywhere — no rows to remove.
- `views` has a standard deviation of only 1% of its mean (9,521 to 10,468).
  It is nearly constant, so it is unlikely to be a useful predictor.
- Mean equals median in every numeric column — perfectly symmetric, no skew.
  Real revenue data is right-skewed, which confirms this dataset is synthetic.

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nAs a percentage of all rows:")
print((df.isnull().sum() / len(df) * 100).round(2))

print("\nExact duplicate rows:", df.duplicated().sum())
```
